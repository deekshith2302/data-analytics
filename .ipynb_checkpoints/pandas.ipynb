```python
# Import Pandas for data manipulation, analysis, and working with DataFrames
import pandas as pd

# Import NumPy for numerical computations and working with arrays
import numpy as np

# Import Seaborn for statistical data visualization
import seaborn as sns

# Import Matplotlib's pyplot module for creating and customizing plots
import matplotlib.pyplot as plt

# Display Matplotlib plots directly inside the Jupyter Notebook
%matplotlib inline

# Set Seaborn plotting configuration and enable Matplotlib color codes
sns.set(color_codes=True)
```

```python
df= pd.read_csv(r'C:\Users\deekshithvarala\data analytics\archive\Automobile.csv')
```

```python
#automatically gives top 5 in the table
df.head()
#head(n) method is used to return the first n rows of a DataFrame.
```

```python
#converts the pandas list to numpy array
df.columns.values
```

```python
#analytic summary of the dataset
df.describe()
```

```python
#histogram using seaborn library

df.hist(figsize=(20,30))
```

```python
#boxplot using sns seaborn
sns.boxplot(x="weight", y="displacement",data=df)
```

```python
sns.boxplot(x='origin', y='horsepower', data=df)
plt.show()
```

```python
sns.boxplot(x='origin', y='mpg', data=df)
plt.show()
```

```python
#pair plot represents the 
sns.pairplot(df[['mpg', 'horsepower','displacement']])
```

```python
sns.pairplot(df[['mpg', 'displacement']])
```

```python
features = [
    'mpg',
    'horsepower',
    'weight',
    'displacement',
    'acceleration'
]

sns.pairplot(df[features])
plt.show()
```

```python
#using catageorical hue values in pairplot
sns.pairplot(
    df[['mpg', 'horsepower', 'weight', 'origin']],
    hue='origin'
)

plt.show()
```

```python
#pairplot using kerenel density engine kde and regression plot reg
sns.pairplot(
    df[['mpg', 'weight', 'horsepower']],
    kind='reg',
    diag_kind='kde'
)
```

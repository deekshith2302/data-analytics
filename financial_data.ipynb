```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
```

```python
sns.set_style('darkgrid')
```

```python
df =pd.read_csv('C:/Users/deekshithvarala/data analytics/finance_data.csv')
```

```python
df.head()
```

```python
#column names or features
print(df.columns.tolist())
```

```python
#droping culumns 
df.drop(['Mutual_Funds','Equity_Market','Debentures','Government_Bonds','PPF','Fixed_Deposits','Gold'],axis =1 ,inplace=True)
```

```python
#replacing whitespaces and null values in string as Nan
df=df.replace(r'^\s*$',np.nan,regex=True)
```

```python
sns.countplot(
    x='gender',
    data=df,
    hue='gender',
    palette='Set2',
    linewidth=3,
    edgecolor='black',
    legend=False
)

plt.show()
```

```python
df['gender'].value_counts()
```

```python
#to calculate the exact percentage from whole 100
df['gender'].value_counts(normalize=True) *100
```

```python
sns.countplot(x='age',
              data=df,
              palette="Set3",
              linewidth=2,
              edgecolor="grey",
              legend=False)
plt.show()
```

```python
plt.figure(figsize=(18,6))
plt.subplot(1,2,1)
sns.countplot(x=df['gender'],hue=df['Investment_Avenues'],
              palette="summer",
              linewidth=3,
              edgecolor="white",
              legend=False)

plt.subplot(1,2,2)
sns.countplot(x=df['gender'],hue=df['Stock_Marktet'],
              palette='hot',
              linewidth=3,
              edgecolor="black")
```

```python
sns.countplot(x=df['Factor'],palette='coolwarm',
              linewidth=2,
              edgecolor='black')
plt.show()
```

```python
plt.figure(figsize=(16,8))
sns.pointplot(x='Purpose',y='age',data=df,
              linestyle= "--",
              capsize=.3,
              color ='Blue')
plt.show()
```

```python
plt.figure(figsize=(16,8))
plt.subplot(1,2,1)
sns.countplot(hue=df['gender'],x=df['Duration']
              ,palette='viridis',
              linewidth=2,edgecolor="black")
plt.title=("DURATION")
plt.subplot(1,2,2)
sns.countplot(hue=df['gender'],x=df['Invest_Monitor'],
              palette='seismic',
              linewidth=3,
              edgecolor='grey')
plt.title("INVESTMENT MONITORING")
plt.show()
```
